```python
import pandas as pd
```

```python
bios = pd.read_csv("./data/bios.csv")
```

```python
bios.head()
```

What should we clean up at first? 

- [x] Get rid of bullet points in the names
- [] Get rid of extra columns
- [] Try to ensure that there is only one piece of information in every column, for example, we should separate the birthday/location in the `Born` and `Die` Column
- [] Parse out city, region, and country from `Born` column
- [] Parse out dates form `Born` & `Die` column

```python
df = bios.copy()
```

```python
df.info()
```

Get rid of bullet points in the dataframe

```python
df['name'] = df['Used name'].str.replace('•', " ")
```

```python
df.head()
```

```python
# We want to split the measurements into two separete columns
df[['height_cm', 'weight_kg']] = df['Measurements'].str.split('/', expand = True)
```

```python
# Check to make sure all the values in Measurements have the '/' symbol in it
df[~df['Measurements'].str.contains('/', na=False) & df['Measurements'].notna()]
# Above will return all the measurements that does not contain /

# Get rid of "cm" and the "kg" from our new columns

# df.head()
```

```python
# some small toy example to infer the overall behaviour on the actual data frame.
small_data = pd.DataFrame(['183 cm', '190cm / 150 kg', '100 kg'], columns = ['Measurements'])
small_data[['height_cm', 'weight_kg']] = small_data['Measurements'].str.split('/', expand = True)

# Strategy, we first populate hte weight_kg column with the values of the height_cm after splitting, this ensures that if weight_kg is NAN, it gets the value of either the height or the actual weight
small_data.fillna({'weight_kg': small_data['height_cm']}, inplace=True)

# Then we strip the unit away
small_data['height_cm'] = pd.to_numeric(small_data['height_cm'].str.strip(' cm'), errors="coerce")
small_data['weight_kg'] = pd.to_numeric(small_data['weight_kg'].str.strip(' kg'), errors="coerce")

small_data.head()
```

```python
df.head()
```

```python
# df['height_cm'] = df['height_cm'].str.extract(r'(\d+)')

# However, we do not want to use this approach, when we observe from the toy example we see above. We see that for 
# weights of 100kg, after splitting, it will also fall under the heights column. So to effectively avoid that situation, 
# we should leverage split instead.
df.head()
```

```python
# After expriementing with the toy dataset, we can apply the same strategy on the column of measurements

# Since we have already divided into height_cm and weight_kg, we now just need to perform copy and unit removals
df.fillna({"weight_kg": df['height_cm']}, inplace=True)
df.head()
```

```python
df[~df["weight_kg"].notna()]  # Wec check and see that all values of weight_kg have been filled with real values
```

```python
df['height_cm'] = pd.to_numeric(df['height_cm'].str.strip(" cm"), errors="coerce")
df['weight_kg'] = pd.to_numeric(df['weight_kg'].str.strip(" kg"), errors="coerce")
```

```python
df.head()
```

```python
# Parse out dates from Born and Die columns
date_pattern = r'(\d+ \w+ \d{4})'

df['Born Date'] = df['Born'].str.extract(date_pattern)
```

```python
df.head()
```

```python
# Build small examples
small_data = pd.DataFrame(['1895 in Scotland', '12 December 1990 in Boston'], columns=['Born'])
dat_pattern = r'(\d+ \w+ \d{4}|\d{4})'
small_data['born_date'] = small_data['Born'].str.extract(date_pattern)
```

```python
small_data
```

```python
date_pattern = r'(\d+ \w+ \d{4}|\d{4})'
df['born_date'] = df['Born'].str.extract(date_pattern)
df['born_date'] = pd.to_datetime(df['born_date'], format="mixed", errors='coerce')
```

```python
df.head()
```

```python
# Get city, region, and country from Born column
location_pattern = r'in ([\w\s]+), ([\w\s]+) \((\w+)\)'
```

```python
df['Born'].str.extract(location_pattern)
```

```python
df[['born_city', 'born_region', 'born_country']] = df['Born'].str.extract(location_pattern, expand = True)
```

```python
df['died_date'] = df['Died'].str.extract(date_pattern)
df['died_date'] = pd.to_datetime(df['died_date'], errors = "coerce")
```

```python
df.head()
df.info()
```
