```python
import pandas as pd
import numpy as np
```

### This file is specifically designed as a practice notebook for exploring different pandas tools

Aggregation

```python
df0 = pd.DataFrame([
    [1, 2, 3], 
    [4, 5, 6],
    [7, 8, 9],
    [np.nan, np.nan, np.nan]], 
columns= ['A', 'B', 'C'])
```

```python
df0.head()
```

```python
df0.agg(['sum', 'min'])  # by default, using aggregate applies on the columns
```

```python
df0.agg({'A': ['sum', 'min'], 'B': ['min', 'max']})
```

```python
df0.agg("mean", axis = 1)  # rowwise aggregation
```

```python
df0.agg(x=('A', 'max'), y=('B', 'min'), z=('C', 'mean'))
```

```python
df0
```

```python
df0.sum()
df0.count()
df0.median()
df0.quantile([0.25, 0.75])
df0.apply(lambda x: x **2)
df0.min()
df0.max()
df0.mean()
df0.var()
df0.std()
```

```python
df0['ranked_densed'] = df0['A'].rank(method = 'dense')
df0['rank_min'] = df0['C'].rank(method='min')
```
