#### Model Implementation from Scratch

##### How the algorithm actually works
We need to caluclate the euclidean distane between a new sample and all points. We determine the lable of the sample based on the majority vote.

##### Key Points:
Euclidean Distance
Euclidean distance is defined as the distance between two points, which is represetend by the equation:<br>
$$\sqrt{\sum_{i = 0}^{m - 1} (x - y)^2}$$

```python
class KNN:

    def __init__(self, n_neighbors = 5):
        self.n_neighbors = n_neighbors
    
    def euclidean_distance(self, x1, x2):
        """
            Calculate the Euclidean distance between 2 data points.
            x1, x2: numpy.ndarray, shape(n_features, )

            Returns: distance, float
        """
        return np.linalg.norm(x1 - x2)
    
    def fit(self, X_train, y_train):
        self.X_train = X_train
        self.y_train = y_train

    def predict(self, X):
        """
            Predicts the class lables for each example in X.
            Parameters:
                X: numpy.ndarray, shape (n_samples, n_features) The test dataset.
            Returns:
                predictsion, shape (n_samples, )
        """
        predictions = []
        for x in X:
            prediction = self._predict(x)
            predictions.append(prediction)
        return np.array(predictions)

    def _predict(self, x):
        distances = []

        # Loop through all the training examples, and compute the distance btween x and all
        # the training examples
        for x_train in self.X_train:
            distance = self.euclidean_distance(x_train, x)
            distances.append(distance)
        distances = np.array(distances)

        # Sort by ascending distance and return indices of the given n neighbors
        n_neighbor_idxs = np.argsort(distances)[: self.n_neighbors]

        # Retrieve the labels/classes of the training examples from the indices
        labels = self.y_train[n_neighbor_idxs]
        labels = list(labels)

        most_occurring_label = max(labels, key= labels.count)
        return most_occurring_label

```

#### Evaluation

```python
def train_test_split(X, y, random_state = 42, test_size = 0.2):
    """
        Splits the data into training and testing sets.
        Parameters:
            X (numpy.ndarray): Features array of shape (n_samples, n_features)
            y (numpy.ndarray): Target array of shape (n_samples, ).
        
        Returns:
            Tuple[numpy.ndarray]: A tuple containing X_train, X_test, y_train, y_test
    """
    # Get number of samples
    n_samples = X.shape[0]
    np.random.seed(random_state)

    # Shuffle the indices:
    shuffled_indices = np.random.permutation(np.arange(n_samples))

    # Determine the size of the test set
    test_size = int(n_samples * test_size)

    # Split the indices into test and train
    test_indices = shuffled_indices[:test_size]
    train_indices = shuffled_indices[test_size:]

    # Split the features and target arrays into test and train
    X_train, X_test = X[train_indices], X[test_indices]
    y_train, y_test = y[train_indices], y[test_indices]

    return X_train, X_test, y_train, y_test    

```

```python
X_train, X_test, y_train, y_test = train_test_split(X.values, y.values, test_size=0.2, random_state=42)
```

```python
model = KNN(7)
model.fit(X_train, y_train)
```

```python
def compute_accuracy(y_true, y_pred):
    y_true = y_true.flatten()
    total_samples = len(y_true)
    correct_predictions = np.sum(y_true == y_pred)
    return (correct_predictions / total_samples)
```

```python
predictions = model.predict(X_test)
accuracy = compute_accuracy(y_test, predictions)
print(f"Our KNN Model fot accuracy score of {accuracy}")
```
