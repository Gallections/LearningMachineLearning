## Using Linear Regression Model from scikit Learn Library

After having implemented linear regression from scratch, it's time to learn to use the scikit learn lirbary for linear regression.

```python
import os
import pandas as pd
import numpy as np
import plotly.express as px
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error
```

##### Data Analysis

```python
train = pd.read_csv('./data/train.csv')
test = pd.read_csv('./data/test.csv')
```

```python
train.head()
```

```python
train.isnull().sum()
```

```python
test.isnull().sum()
```

##### Data Preprocessing

```python
# Drop null values
train = train.dropna()
```

```python
# Set training data and targets
X_train = train['x']
y_train = train['y']

# Set testing data and targets
X_test = test['x']
y_test = test['y']

print(y_test.shape)
print(y_train.shape)

print(X_train.shape)
print(X_test.shape)
```

!! Be careful, we need to reshape the data in order to fit our model

```python
X_train = X_train.values.reshape(-1, 1)
X_test = X_test.values.reshape(-1,1)
```

```python
scaler = StandardScaler()
scaler.fit(X_train)

X_train = scaler.transform(X_train)
X_test = scaler.transform(X_test)
print("Transformation successful")
```

```python
X_train.min(), X_train.max()
```

##### Visualize The Data

```python
px.scatter(x= train['x'], y=train['y'], template = 'gridon')
```

#### Model

```python
model = LinearRegression() # Create linear regression instance

model.fit(X_train, y_train)
```

##### Evaluation

```python
predictions = model.predict(X_test)
```

```python
mse = mean_squared_error(y_test, predictions)
```

```python
print(f'the mean squared error is: {mse}')
```
